# Exploratory Data Analysis (EDA) Project
## Sales Data Analysis

**Objective:** Analyze a sales dataset to uncover patterns, trends, correlations, and key influencing factors using statistical summaries and visualizations.

### Tools Used
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Jupyter Notebook


## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
plt.rcParams["figure.figsize"] = (9, 5)
```

## 2. Load the Dataset

```python
df = pd.read_csv("raw_sales_dataset.csv")
df.head()
```

## 3. Understand the Dataset

```python
print("Rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())
```

## 4. Check Missing Values and Duplicates

```python
print("Missing values:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:", df.duplicated().sum())
```

## 5. Statistical Summary

```python
df.describe(include="all")
```

## 6. Data Cleaning

```python
# Convert date column
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Fill categorical missing values with the mode
for col in ["Region", "Category"]:
    df[col] = df[col].fillna(df[col].mode()[0])

# Fill numeric missing values with the median
for col in ["Sales", "Quantity"]:
    df[col] = df[col].fillna(df[col].median())

# Remove duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

print("Missing values after cleaning:")
print(df.isnull().sum())
print("\nDuplicates after cleaning:", df.duplicated().sum())
```

## 7. Feature Engineering

```python
df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.strftime("%b")

monthly_sales = df.groupby("Month")["Sales"].sum()
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

monthly_sales.head(), category_sales, region_sales
```

## 8. Univariate Analysis – Sales Distribution

```python
plt.figure()
plt.hist(df["Sales"], bins=30)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

## 9. Category-wise Sales

```python
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

## 10. Region-wise Sales

```python
region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

## 11. Monthly Sales Trend

```python
plt.figure()
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13))
plt.grid(True)
plt.show()
```

## 12. Quantity Distribution by Category

```python
plt.figure()
sns.boxplot(data=df, x="Category", y="Quantity")
plt.title("Quantity Distribution by Category")
plt.xticks(rotation=15)
plt.show()
```

## 13. Sales vs Quantity

```python
plt.figure()
sns.scatterplot(data=df, x="Quantity", y="Sales", hue="Category")
plt.title("Sales vs Quantity")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.show()
```

## 14. Correlation Heatmap

```python
numeric_cols = ["Quantity", "Unit_Price", "Discount", "Sales"]
corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## 15. Identify Potential Outliers

```python
plt.figure()
sns.boxplot(y=df["Sales"])
plt.title("Sales Outlier Detection")
plt.ylabel("Sales")
plt.show()

plt.figure()
sns.boxplot(y=df["Quantity"])
plt.title("Quantity Outlier Detection")
plt.ylabel("Quantity")
plt.show()
```

## 16. Key Business Insights

```python
print("Total Sales:", round(df["Sales"].sum(), 2))
print("Average Sales per Order:", round(df["Sales"].mean(), 2))
print("Top Category:", category_sales.idxmax())
print("Top Region:", region_sales.idxmax())
print("Highest Sales Month:", monthly_sales.idxmax())

print("\nCategory-wise sales:")
display(category_sales)

print("\nRegion-wise sales:")
display(region_sales)
```

## 17. Findings

1. Sales vary across product categories, showing that some categories contribute more revenue than others.
2. Regional analysis helps identify areas with higher sales activity.
3. Monthly analysis shows how sales change throughout the year and helps identify high- and low-performing months.
4. The scatter plot helps examine the relationship between quantity sold and sales revenue.
5. The correlation heatmap shows relationships among quantity, unit price, discount, and sales.
6. Box plots help identify unusually high or low observations that may influence statistical analysis.
7. Data cleaning was necessary because the raw dataset contained missing values and duplicate records.
